```python
from ultralytics import YOLO
```

```python
model=YOLO('yolov8n-cls.pt')
```

```python
model.train(data="/content/drive/MyDrive/ML dataset/dataset",epochs=30,imgsz=224,batch=16)
```

```python
metrics= model.val()
print(metrics)
```

```python
results=model.predict("/content/drive/MyDrive/ML dataset/dataset/val/hsrp/0CKYHOL_lgmnkpq7_1_47383079.jpg")
print(results[0].probs)
```
